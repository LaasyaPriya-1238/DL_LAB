```python
# ============================================================
# SECTIONS 18-21
# VIDEO UNDERSTANDING USING CNN + RNN
# PART 1: SETUP + DATASET + FRAME EXTRACTION
# ============================================================

import os
import cv2
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Input,
    Dense,
    SimpleRNN,
    LSTM,
    GRU,
    GlobalAveragePooling2D
)

from tensorflow.keras.applications import MobileNetV2

from tensorflow.keras.applications.mobilenet_v2 import (
    preprocess_input
)

from tensorflow.keras.callbacks import EarlyStopping


# ============================================================
# RANDOM SEEDS
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# ============================================================
# UCF101 DATASET PATH
# ============================================================
#
# Your extracted dataset is already here:
#
# /content/UCF101/UCF-101/
#
# So we DO NOT download or extract anything again.
# ============================================================

UCF101_PATH = "/content/UCF101/UCF-101"


# ============================================================
# VERIFY DATASET PATH
# ============================================================

if not os.path.exists(UCF101_PATH):

    raise FileNotFoundError(
        f"UCF101 dataset not found at:\n{UCF101_PATH}"
    )


print("========================================")
print("UCF101 DATASET")
print("========================================")

print(
    "Dataset path:",
    UCF101_PATH
)


# ============================================================
# SELECTED ACTION CLASSES
# ============================================================
#
# We are using 3 classes because these three were confirmed
# to exist in your extracted UCF101 dataset.
#
# The assignment requires 3-5 classes, so 3 is valid.
# ============================================================

SELECTED_CLASSES = [

    "Basketball",

    "Biking",

    "TennisSwing"

]


# ============================================================
# PARAMETERS
# ============================================================

MAX_VIDEOS_PER_CLASS = 20

NUM_FRAMES = 10

IMG_SIZE = 224

RNN_UNITS = 32

EPOCHS = 30

BATCH_SIZE = 8

NUM_CLASSES = len(
    SELECTED_CLASSES
)


# ============================================================
# VERIFY SELECTED CLASS FOLDERS
# ============================================================

print("\nSelected classes:")

for class_name in SELECTED_CLASSES:

    class_folder = os.path.join(

        UCF101_PATH,

        class_name

    )


    if not os.path.isdir(
        class_folder
    ):

        raise FileNotFoundError(

            f"Class folder not found:\n"
            f"{class_folder}"

        )


    videos = [

        f

        for f in os.listdir(
            class_folder
        )

        if f.lower().endswith(

            (
                ".avi",
                ".mp4",
                ".mov"
            )

        )

    ]


    print(
        f"✓ {class_name}: "
        f"{len(videos)} videos"
    )


# ============================================================
# CLASS LABEL MAPPING
# ============================================================

class_to_index = {

    class_name: index

    for index, class_name
    in enumerate(
        SELECTED_CLASSES
    )

}


index_to_class = {

    index: class_name

    for class_name, index
    in class_to_index.items()

}


print("\nClass mapping:")

print(
    class_to_index
)


# ============================================================
# COLLECT VIDEO PATHS
# ============================================================

video_paths = []

labels = []


for class_name in SELECTED_CLASSES:

    class_folder = os.path.join(

        UCF101_PATH,

        class_name

    )


    videos = [

        os.path.join(
            class_folder,
            filename
        )

        for filename
        in os.listdir(
            class_folder
        )

        if filename.lower().endswith(

            (
                ".avi",
                ".mp4",
                ".mov"
            )

        )

    ]


    # Sort for reproducibility
    videos = sorted(
        videos
    )


    # Use only a small subset
    # as required by the experiment
    videos = videos[
        :MAX_VIDEOS_PER_CLASS
    ]


    for video in videos:

        video_paths.append(
            video
        )

        labels.append(
            class_to_index[
                class_name
            ]
        )


# ============================================================
# DATASET SUMMARY
# ============================================================

print(
    "\n========================================"
)

print(
    "VIDEO DATASET SUMMARY"
)

print(
    "========================================"
)


print(
    "Total videos:",
    len(video_paths)
)


for class_name in SELECTED_CLASSES:

    class_index = class_to_index[
        class_name
    ]


    count = sum(

        1

        for label in labels

        if label == class_index

    )


    print(

        f"{class_name}: "
        f"{count} videos"

    )


# ============================================================
# TRAIN / TEST SPLIT
# ============================================================
#
# 80% training
# 20% testing
#
# Stratified split keeps all 3 classes represented.
# ============================================================

video_train, video_test, y_train, y_test = (

    train_test_split(

        video_paths,

        labels,

        test_size=0.20,

        random_state=SEED,

        stratify=labels

    )

)


print(
    "\n========================================"
)

print(
    "TRAIN / TEST SPLIT"
)

print(
    "========================================"
)


print(
    "Training videos:",
    len(video_train)
)


print(
    "Testing videos:",
    len(video_test)
)


# ============================================================
# UNIFORM VIDEO FRAME SAMPLING
# ============================================================
#
# Requirement:
# Sample 10 frames uniformly from each video.
#
# Each frame:
# 224 × 224 × 3
# ============================================================

def sample_video_frames(

    video_path,

    num_frames=NUM_FRAMES

):


    # Open video
    cap = cv2.VideoCapture(
        video_path
    )


    # Number of frames
    total_frames = int(

        cap.get(

            cv2.CAP_PROP_FRAME_COUNT

        )

    )


    # Check video
    if total_frames <= 0:

        cap.release()

        return None


    # --------------------------------------------------------
    # Uniformly select frame positions
    # --------------------------------------------------------

    frame_indices = np.linspace(

        0,

        total_frames - 1,

        num_frames

    ).astype(int)


    frames = []


    # --------------------------------------------------------
    # Read selected frames
    # --------------------------------------------------------

    for index in frame_indices:


        cap.set(

            cv2.CAP_PROP_POS_FRAMES,

            int(index)

        )


        success, frame = cap.read()


        if not success:

            continue


        # OpenCV uses BGR
        # Convert to RGB
        frame = cv2.cvtColor(

            frame,

            cv2.COLOR_BGR2RGB

        )


        # Resize to 224 × 224
        frame = cv2.resize(

            frame,

            (
                IMG_SIZE,
                IMG_SIZE
            )

        )


        frames.append(
            frame
        )


    cap.release()


    # --------------------------------------------------------
    # Check whether frames were obtained
    # --------------------------------------------------------

    if len(frames) == 0:

        return None


    # --------------------------------------------------------
    # Ensure exactly 10 frames
    # --------------------------------------------------------

    while len(frames) < num_frames:

        frames.append(

            frames[-1].copy()

        )


    return np.array(
        frames
    )


# ============================================================
# TEST FRAME EXTRACTION
# ============================================================

sample_video = video_train[0]


sample_frames = sample_video_frames(

    sample_video,

    NUM_FRAMES

)


print(
    "\n========================================"
)

print(
    "FRAME EXTRACTION TEST"
)

print(
    "========================================"
)


print(
    "Sample video:",
    os.path.basename(
        sample_video
    )
)


if sample_frames is not None:

    print(
        "Sampled frame tensor shape:",
        sample_frames.shape
    )

    print(
        "Expected:",
        "(10, 224, 224, 3)"
    )

else:

    raise RuntimeError(
        "Could not extract frames from sample video."
    )


# ============================================================
# FINAL CHECK
# ============================================================

print(
    "\n========================================"
)

print(
    "PART 1 COMPLETED SUCCESSFULLY"
)

print(
    "========================================"
)

print(
    "Classes:",
    SELECTED_CLASSES
)

print(
    "Videos:",
    len(video_paths)
)

print(
    "Frames per video:",
    NUM_FRAMES
)

print(
    "Frame size:",
    f"{IMG_SIZE} × {IMG_SIZE} × 3"
)

print(
    "Sample frame tensor:",
    sample_frames.shape
)
```

```python
# ============================================================
# SECTION 20
# CNN FEATURE EXTRACTION
# PART 2: MobileNetV2 + Plot 7
# ============================================================

# ============================================================
# LOAD PRETRAINED MOBILENETV2
# ============================================================

base_cnn = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze pretrained CNN
base_cnn.trainable = False


# ============================================================
# GLOBAL AVERAGE POOLING
# ============================================================

feature_extractor = Sequential([

    base_cnn,

    GlobalAveragePooling2D()

])


# ============================================================
# CNN FEATURE DIMENSION
# ============================================================

D = feature_extractor.output_shape[-1]

print("========================================")
print("CNN FEATURE EXTRACTOR")
print("========================================")

print(
    "CNN Feature Dimension D:",
    D
)


# ============================================================
# PLOT 7: VIDEO SAMPLE FRAMES
# ============================================================

plt.figure(
    figsize=(15, 6)
)

for i in range(NUM_FRAMES):

    plt.subplot(
        2,
        5,
        i + 1
    )

    plt.imshow(
        sample_frames[i]
    )

    plt.title(
        f"Frame {i + 1}"
    )

    plt.axis("off")


plt.suptitle(
    "Plot 7: Uniformly Sampled Frames from a UCF101 Video",
    fontsize=15
)

plt.tight_layout()

plt.show()


# ============================================================
# FEATURE EXTRACTION FUNCTION
# ============================================================

def extract_video_features(video_path):

    # Get 10 frames
    frames = sample_video_frames(
        video_path,
        NUM_FRAMES
    )

    if frames is None:

        return None


    # Convert to float32
    frames = frames.astype(
        np.float32
    )


    # MobileNetV2 preprocessing
    frames = preprocess_input(
        frames
    )


    # Extract CNN features
    features = feature_extractor.predict(
        frames,
        verbose=0
    )


    return features


# ============================================================
# TEST ONE VIDEO
# ============================================================

test_features = extract_video_features(
    sample_video
)


print("\n========================================")
print("SINGLE VIDEO FEATURE TEST")
print("========================================")

print(
    "Original video frames:",
    sample_frames.shape
)

print(
    "CNN feature shape:",
    test_features.shape
)

print(
    "Expected:",
    f"(10, {D})"
)


# ============================================================
# CREATE FEATURE DATASET
# ============================================================

def create_feature_dataset(
    videos,
    labels
):

    X = []

    y = []

    failed_videos = []


    for i, (video, label) in enumerate(
        zip(videos, labels)
    ):

        print(
            f"[{i + 1}/{len(videos)}] "
            f"Processing: "
            f"{os.path.basename(video)}"
        )


        features = extract_video_features(
            video
        )


        if features is None:

            failed_videos.append(
                video
            )

            continue


        X.append(
            features
        )

        y.append(
            label
        )


    X = np.array(
        X,
        dtype=np.float32
    )

    y = np.array(
        y,
        dtype=np.int32
    )


    print(
        "\nSuccessful videos:",
        len(X)
    )

    print(
        "Failed videos:",
        len(failed_videos)
    )


    return X, y, failed_videos


# ============================================================
# TRAINING FEATURES
# ============================================================

print(
    "\n========================================"
)

print(
    "EXTRACTING TRAINING FEATURES"
)

print(
    "========================================"
)


X_train, y_train, failed_train = (
    create_feature_dataset(
        video_train,
        y_train
    )
)


# ============================================================
# TESTING FEATURES
# ============================================================

print(
    "\n========================================"
)

print(
    "EXTRACTING TESTING FEATURES"
)

print(
    "========================================"
)


X_test, y_test, failed_test = (
    create_feature_dataset(
        video_test,
        y_test
    )
)


# ============================================================
# REQUIRED OBSERVATIONS
# ============================================================

print(
    "\n========================================"
)

print(
    "CNN FEATURE EXTRACTION RESULTS"
)

print(
    "========================================"
)

print(
    "CNN feature dimension D:",
    D
)

print(
    "Training tensor shape:",
    X_train.shape
)

print(
    "Testing tensor shape:",
    X_test.shape
)


print(
    "\nRequired representation:"
)

print(
    f"One video = 10 × {D}"
)

print(
    f"Batch = B × 10 × {D}"
)


# ============================================================
# SAVE FEATURES
# ============================================================

FEATURE_FILE = (
    "/content/"
    "UCF101_MobileNetV2_features.npz"
)


np.savez_compressed(

    FEATURE_FILE,

    X_train=X_train,

    y_train=y_train,

    X_test=X_test,

    y_test=y_test

)


print(
    "\nFeatures saved to:"
)

print(
    FEATURE_FILE
)
```

```python
# ============================================================
# PART 3 — RNN, LSTM AND GRU CLASSIFIERS
# ============================================================

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, SimpleRNN, LSTM, GRU, Dense
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("=" * 70)
print("PART 3 — RNN, LSTM AND GRU")
print("=" * 70)

# ------------------------------------------------------------
# 1. CHECK DATA
# ------------------------------------------------------------

print("\nInput data:")
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape :", X_test.shape)
print("y_test shape :", y_test.shape)

TIME_STEPS = X_train.shape[1]      # 10
FEATURE_DIM = X_train.shape[2]     # 1280
NUM_CLASSES = len(SELECTED_CLASSES)
RNN_UNITS = 32

print("\nTime steps :", TIME_STEPS)
print("Feature dimension :", FEATURE_DIM)
print("Number of classes :", NUM_CLASSES)
print("Recurrent units :", RNN_UNITS)


# ------------------------------------------------------------
# 2. FUNCTION TO CREATE MODELS
# ------------------------------------------------------------

def create_rnn_model():
    model = Sequential([
        Input(shape=(TIME_STEPS, FEATURE_DIM)),
        SimpleRNN(RNN_UNITS, activation="tanh"),
        Dense(NUM_CLASSES, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


def create_lstm_model():
    model = Sequential([
        Input(shape=(TIME_STEPS, FEATURE_DIM)),
        LSTM(RNN_UNITS),
        Dense(NUM_CLASSES, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


def create_gru_model():
    model = Sequential([
        Input(shape=(TIME_STEPS, FEATURE_DIM)),
        GRU(RNN_UNITS),
        Dense(NUM_CLASSES, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ------------------------------------------------------------
# 3. TRAINING FUNCTION
# ------------------------------------------------------------

def train_and_evaluate(model, model_name):

    print("\n" + "=" * 70)
    print("TRAINING:", model_name)
    print("=" * 70)

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    )

    history = model.fit(
        X_train,
        y_train,
        validation_split=0.2,
        epochs=30,
        batch_size=8,
        callbacks=[early_stop],
        verbose=1
    )

    # --------------------------------------------------------
    # TEST PREDICTIONS
    # --------------------------------------------------------

    y_prob = model.predict(X_test, verbose=0)

    y_pred = np.argmax(y_prob, axis=1)

    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    parameters = model.count_params()

    print("\n" + "-" * 50)
    print(model_name, "RESULTS")
    print("-" * 50)

    print(f"Accuracy  : {accuracy:.4f}")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1-Score  : {f1:.4f}")
    print(f"Parameters: {parameters:,}")

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "Parameters": parameters,
        "History": history,
        "Predictions": y_pred,
        "Probabilities": y_prob
    }


# ------------------------------------------------------------
# 4. CREATE MODELS
# ------------------------------------------------------------

rnn_model = create_rnn_model()
lstm_model = create_lstm_model()
gru_model = create_gru_model()

print("\nMODEL ARCHITECTURES")

print("\n--- Simple RNN ---")
rnn_model.summary()

print("\n--- LSTM ---")
lstm_model.summary()

print("\n--- GRU ---")
gru_model.summary()


# ------------------------------------------------------------
# 5. TRAIN ALL THREE MODELS
# ------------------------------------------------------------

rnn_results = train_and_evaluate(
    rnn_model,
    "Simple RNN"
)

lstm_results = train_and_evaluate(
    lstm_model,
    "LSTM"
)

gru_results = train_and_evaluate(
    gru_model,
    "GRU"
)


# ------------------------------------------------------------
# 6. CONSOLIDATED RESULTS TABLE
# ------------------------------------------------------------

results_table = pd.DataFrame([
    {
        "Model": rnn_results["Model"],
        "Accuracy": rnn_results["Accuracy"],
        "Precision": rnn_results["Precision"],
        "Recall": rnn_results["Recall"],
        "F1-Score": rnn_results["F1-Score"],
        "Parameters": rnn_results["Parameters"]
    },
    {
        "Model": lstm_results["Model"],
        "Accuracy": lstm_results["Accuracy"],
        "Precision": lstm_results["Precision"],
        "Recall": lstm_results["Recall"],
        "F1-Score": lstm_results["F1-Score"],
        "Parameters": lstm_results["Parameters"]
    },
    {
        "Model": gru_results["Model"],
        "Accuracy": gru_results["Accuracy"],
        "Precision": gru_results["Precision"],
        "Recall": gru_results["Recall"],
        "F1-Score": gru_results["F1-Score"],
        "Parameters": gru_results["Parameters"]
    }
])

print("\n" + "=" * 70)
print("RNN / LSTM / GRU FINAL RESULTS")
print("=" * 70)

display(
    results_table.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "Parameters": "{:,}"
    })
)


# ------------------------------------------------------------
# 7. TRAINING CURVES
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy
axes[0].plot(
    rnn_results["History"].history["accuracy"],
    label="RNN Train"
)
axes[0].plot(
    rnn_results["History"].history["val_accuracy"],
    label="RNN Validation"
)

axes[0].plot(
    lstm_results["History"].history["accuracy"],
    label="LSTM Train"
)
axes[0].plot(
    lstm_results["History"].history["val_accuracy"],
    label="LSTM Validation"
)

axes[0].plot(
    gru_results["History"].history["accuracy"],
    label="GRU Train"
)
axes[0].plot(
    gru_results["History"].history["val_accuracy"],
    label="GRU Validation"
)

axes[0].set_title("Training and Validation Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")
axes[0].legend()
axes[0].grid(True)

# Loss
axes[1].plot(
    rnn_results["History"].history["loss"],
    label="RNN Train"
)
axes[1].plot(
    rnn_results["History"].history["val_loss"],
    label="RNN Validation"
)

axes[1].plot(
    lstm_results["History"].history["loss"],
    label="LSTM Train"
)
axes[1].plot(
    lstm_results["History"].history["val_loss"],
    label="LSTM Validation"
)

axes[1].plot(
    gru_results["History"].history["loss"],
    label="GRU Train"
)
axes[1].plot(
    gru_results["History"].history["val_loss"],
    label="GRU Validation"
)

axes[1].set_title("Training and Validation Loss")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. CLASSIFICATION REPORTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CLASSIFICATION REPORTS")
print("=" * 70)

for name, result in [
    ("Simple RNN", rnn_results),
    ("LSTM", lstm_results),
    ("GRU", gru_results)
]:

    print("\n" + "-" * 60)
    print(name)
    print("-" * 60)

    print(
        classification_report(
            y_test,
            result["Predictions"],
            target_names=SELECTED_CLASSES,
            zero_division=0
        )
    )


# ------------------------------------------------------------
# 9. SAVE RESULTS
# ------------------------------------------------------------

results_table.to_csv(
    "/content/RNN_LSTM_GRU_results.csv",
    index=False
)

print("\nResults saved to:")
print("/content/RNN_LSTM_GRU_results.csv")

print("\n" + "=" * 70)
print("PART 3 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# PART 5 — SEQUENCE-TO-SEQUENCE REVERSAL USING LSTM
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    LSTM,
    Dense
)
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split

print("=" * 70)
print("PART 5 — SEQUENCE-TO-SEQUENCE REVERSAL")
print("=" * 70)


# ============================================================
# 1. PARAMETERS
# ============================================================

NUM_SEQUENCES = 5000
SEQUENCE_LENGTH = 4
VOCAB_SIZE = 10
LATENT_DIM = 64

EPOCHS = 30
BATCH_SIZE = 64

START_TOKEN = VOCAB_SIZE
TOTAL_VOCAB_SIZE = VOCAB_SIZE + 1

np.random.seed(42)


# ============================================================
# 2. GENERATE SYNTHETIC SEQUENCES
# ============================================================

# Example:
# Input  : [1, 4, 7, 2]
# Target : [2, 7, 4, 1]

X = np.random.randint(
    0,
    VOCAB_SIZE,
    size=(NUM_SEQUENCES, SEQUENCE_LENGTH)
)

Y = np.flip(X, axis=1)

print("\nGenerated data:")
print("Input shape :", X.shape)
print("Target shape:", Y.shape)

print("\nExample sequences:")

for i in range(5):
    print(
        f"Input: {X[i].tolist()}  ->  "
        f"Output: {Y[i].tolist()}"
    )


# ============================================================
# 3. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

X_temp, X_test, Y_temp, Y_test = train_test_split(
    X,
    Y,
    test_size=0.20,
    random_state=42
)

X_train, X_val, Y_train, Y_val = train_test_split(
    X_temp,
    Y_temp,
    test_size=0.20,
    random_state=42
)

print("\nDataset split:")
print("Training   :", X_train.shape[0])
print("Validation :", X_val.shape[0])
print("Testing    :", X_test.shape[0])


# ============================================================
# 4. ONE-HOT ENCODING
# ============================================================

# Encoder input
encoder_input_data = to_categorical(
    X_train,
    num_classes=VOCAB_SIZE
)

encoder_val_data = to_categorical(
    X_val,
    num_classes=VOCAB_SIZE
)

encoder_test_data = to_categorical(
    X_test,
    num_classes=VOCAB_SIZE
)


# ------------------------------------------------------------
# Decoder uses teacher forcing
#
# Example target:
# [2, 7, 4, 1]
#
# Decoder input:
# [START, 2, 7, 4]
#
# Decoder target:
# [2, 7, 4, 1]
# ------------------------------------------------------------

decoder_input_train = np.zeros(
    (len(Y_train), SEQUENCE_LENGTH),
    dtype=int
)

decoder_input_val = np.zeros(
    (len(Y_val), SEQUENCE_LENGTH),
    dtype=int
)

decoder_input_test = np.zeros(
    (len(Y_test), SEQUENCE_LENGTH),
    dtype=int
)

decoder_input_train[:, 0] = START_TOKEN
decoder_input_train[:, 1:] = Y_train[:, :-1]

decoder_input_val[:, 0] = START_TOKEN
decoder_input_val[:, 1:] = Y_val[:, :-1]

decoder_input_test[:, 0] = START_TOKEN
decoder_input_test[:, 1:] = Y_test[:, :-1]


decoder_input_train = to_categorical(
    decoder_input_train,
    num_classes=TOTAL_VOCAB_SIZE
)

decoder_input_val = to_categorical(
    decoder_input_val,
    num_classes=TOTAL_VOCAB_SIZE
)

decoder_input_test = to_categorical(
    decoder_input_test,
    num_classes=TOTAL_VOCAB_SIZE
)


decoder_target_train = to_categorical(
    Y_train,
    num_classes=VOCAB_SIZE
)

decoder_target_val = to_categorical(
    Y_val,
    num_classes=VOCAB_SIZE
)

decoder_target_test = to_categorical(
    Y_test,
    num_classes=VOCAB_SIZE
)


# ============================================================
# 5. BUILD ENCODER
# ============================================================

encoder_inputs = Input(
    shape=(SEQUENCE_LENGTH, VOCAB_SIZE),
    name="encoder_inputs"
)

encoder_lstm = LSTM(
    LATENT_DIM,
    return_state=True,
    name="encoder_lstm"
)

encoder_outputs, state_h, state_c = encoder_lstm(
    encoder_inputs
)

encoder_states = [state_h, state_c]


# ============================================================
# 6. BUILD DECODER
# ============================================================

decoder_inputs = Input(
    shape=(SEQUENCE_LENGTH, TOTAL_VOCAB_SIZE),
    name="decoder_inputs"
)

decoder_lstm = LSTM(
    LATENT_DIM,
    return_sequences=True,
    return_state=True,
    name="decoder_lstm"
)

decoder_outputs, _, _ = decoder_lstm(
    decoder_inputs,
    initial_state=encoder_states
)

decoder_dense = Dense(
    VOCAB_SIZE,
    activation="softmax",
    name="decoder_output"
)

decoder_outputs = decoder_dense(
    decoder_outputs
)


# ============================================================
# 7. COMPLETE SEQ2SEQ MODEL
# ============================================================

seq2seq_model = Model(
    [encoder_inputs, decoder_inputs],
    decoder_outputs
)

seq2seq_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("\n" + "=" * 70)
print("SEQ2SEQ MODEL")
print("=" * 70)

seq2seq_model.summary()


# ============================================================
# 8. TRAIN MODEL
# ============================================================

print("\n" + "=" * 70)
print("TRAINING SEQ2SEQ MODEL")
print("=" * 70)

history_seq2seq = seq2seq_model.fit(
    [
        encoder_input_data,
        decoder_input_train
    ],
    decoder_target_train,

    validation_data=(
        [
            encoder_val_data,
            decoder_input_val
        ],
        decoder_target_val
    ),

    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    verbose=1
)


# ============================================================
# 9. TEST LOSS AND TOKEN ACCURACY
# ============================================================

test_loss, test_token_accuracy = seq2seq_model.evaluate(
    [
        encoder_test_data,
        decoder_input_test
    ],
    decoder_target_test,
    verbose=0
)

print("\n" + "=" * 70)
print("SEQ2SEQ TEST RESULTS")
print("=" * 70)

print(f"Test Loss        : {test_loss:.4f}")
print(f"Token Accuracy   : {test_token_accuracy:.4f}")


# ============================================================
# 10. PREDICTIONS
# ============================================================

test_predictions = seq2seq_model.predict(
    [
        encoder_test_data,
        decoder_input_test
    ],
    verbose=0
)

predicted_sequences = np.argmax(
    test_predictions,
    axis=-1
)


# ============================================================
# 11. SEQUENCE ACCURACY
# ============================================================

sequence_correct = np.all(
    predicted_sequences == Y_test,
    axis=1
)

sequence_accuracy = np.mean(
    sequence_correct
)

print(f"Sequence Accuracy: {sequence_accuracy:.4f}")


# ============================================================
# 12. DISPLAY 5 TEST EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("5 TEST EXAMPLES")
print("=" * 70)

for i in range(5):

    print(f"\nExample {i + 1}")

    print(
        "Input sequence :",
        X_test[i].tolist()
    )

    print(
        "Expected output:",
        Y_test[i].tolist()
    )

    print(
        "Predicted output:",
        predicted_sequences[i].tolist()
    )

    if sequence_correct[i]:
        print("Result         : CORRECT")
    else:
        print("Result         : INCORRECT")


# ============================================================
# 13. TOKEN-LEVEL ACCURACY MANUALLY
# ============================================================

total_tokens = Y_test.size

correct_tokens = np.sum(
    predicted_sequences == Y_test
)

manual_token_accuracy = (
    correct_tokens / total_tokens
)

print("\n" + "=" * 70)
print("DETAILED ACCURACY")
print("=" * 70)

print(f"Correct tokens : {correct_tokens}")
print(f"Total tokens   : {total_tokens}")
print(f"Token accuracy : {manual_token_accuracy:.4f}")
print(f"Sequence accuracy: {sequence_accuracy:.4f}")


# ============================================================
# 14. TRAINING / VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history_seq2seq.history["loss"],
    label="Training Loss"
)

plt.plot(
    history_seq2seq.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "Seq2Seq Encoder-Decoder LSTM Training and Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid(True)

plt.tight_layout()

plt.savefig(
    "/content/Seq2Seq_Training_Validation_Loss.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 15. FINAL SEQ2SEQ RESULTS TABLE
# ============================================================

seq2seq_results = pd.DataFrame({
    "Metric": [
        "Test Loss",
        "Token Accuracy",
        "Sequence Accuracy"
    ],

    "Value": [
        test_loss,
        manual_token_accuracy,
        sequence_accuracy
    ]
})

print("\n" + "=" * 70)
print("SEQ2SEQ FINAL RESULTS")
print("=" * 70)

display(
    seq2seq_results.style.format({
        "Value": "{:.4f}"
    })
)

print("\nSeq2Seq results saved in variable: seq2seq_results")

print("\n" + "=" * 70)
print("PART 5 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# PART 6 — CNN-LSTM AND CNN-GRU
# CORRECTED VERSION
# ============================================================

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, GRU, Dense
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("=" * 70)
print("PART 6 — CNN-LSTM AND CNN-GRU")
print("=" * 70)


# ============================================================
# 1. RELOAD SAVED CNN FEATURES
# ============================================================

feature_file = "/content/UCF101_MobileNetV2_features.npz"

feature_data = np.load(feature_file)

video_X_train = feature_data["X_train"]
video_y_train = feature_data["y_train"]

video_X_test = feature_data["X_test"]
video_y_test = feature_data["y_test"]


print("\nReloaded MobileNetV2 features:")

print("Training features:", video_X_train.shape)
print("Training labels  :", video_y_train.shape)

print("Testing features :", video_X_test.shape)
print("Testing labels   :", video_y_test.shape)


# ============================================================
# 2. INPUT INFORMATION
# ============================================================

TIME_STEPS = video_X_train.shape[1]
FEATURE_DIM = video_X_train.shape[2]
NUM_CLASSES = len(SELECTED_CLASSES)
RNN_UNITS = 32

print("\nInput representation:")
print("Time steps        :", TIME_STEPS)
print("CNN feature size  :", FEATURE_DIM)
print("Number of classes :", NUM_CLASSES)
print("Recurrent units   :", RNN_UNITS)

print("\nExpected:")
print("Time steps = 10")
print("Feature dimension = 1280")


# ============================================================
# 3. CNN-LSTM MODEL
# ============================================================

cnn_lstm_model = Sequential([
    Input(
        shape=(TIME_STEPS, FEATURE_DIM)
    ),

    LSTM(
        RNN_UNITS,
        activation="tanh"
    ),

    Dense(
        NUM_CLASSES,
        activation="softmax"
    )
])

cnn_lstm_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 4. CNN-GRU MODEL
# ============================================================

cnn_gru_model = Sequential([
    Input(
        shape=(TIME_STEPS, FEATURE_DIM)
    ),

    GRU(
        RNN_UNITS,
        activation="tanh"
    ),

    Dense(
        NUM_CLASSES,
        activation="softmax"
    )
])

cnn_gru_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 5. MODEL SUMMARIES
# ============================================================

print("\n" + "=" * 70)
print("CNN-LSTM ARCHITECTURE")
print("=" * 70)

cnn_lstm_model.summary()


print("\n" + "=" * 70)
print("CNN-GRU ARCHITECTURE")
print("=" * 70)

cnn_gru_model.summary()


# ============================================================
# 6. TRAINING FUNCTION
# ============================================================

def train_cnn_recurrent_model(
    model,
    model_name
):

    print("\n" + "=" * 70)
    print("TRAINING:", model_name)
    print("=" * 70)

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    )

    history = model.fit(
        video_X_train,
        video_y_train,

        validation_split=0.2,

        epochs=30,
        batch_size=8,

        callbacks=[early_stop],

        verbose=1
    )

    # --------------------------------------------------------
    # TEST PREDICTIONS
    # --------------------------------------------------------

    probabilities = model.predict(
        video_X_test,
        verbose=0
    )

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------

    accuracy = accuracy_score(
        video_y_test,
        predictions
    )

    precision = precision_score(
        video_y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        video_y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        video_y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    parameters = model.count_params()

    print("\n" + "-" * 50)
    print(model_name, "RESULTS")
    print("-" * 50)

    print(f"Accuracy  : {accuracy:.4f}")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1-Score  : {f1:.4f}")
    print(f"Parameters: {parameters:,}")

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "Parameters": parameters,
        "History": history,
        "Predictions": predictions,
        "Probabilities": probabilities
    }


# ============================================================
# 7. TRAIN CNN-LSTM
# ============================================================

cnn_lstm_results = train_cnn_recurrent_model(
    cnn_lstm_model,
    "CNN-LSTM"
)


# ============================================================
# 8. TRAIN CNN-GRU
# ============================================================

cnn_gru_results = train_cnn_recurrent_model(
    cnn_gru_model,
    "CNN-GRU"
)


# ============================================================
# 9. CLASSIFICATION REPORTS
# ============================================================

print("\n" + "=" * 70)
print("CNN-LSTM CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        video_y_test,
        cnn_lstm_results["Predictions"],
        target_names=SELECTED_CLASSES,
        zero_division=0
    )
)


print("\n" + "=" * 70)
print("CNN-GRU CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        video_y_test,
        cnn_gru_results["Predictions"],
        target_names=SELECTED_CLASSES,
        zero_division=0
    )
)


# ============================================================
# 10. TRAINING CURVES
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5)
)


# -----------------------------
# Accuracy
# -----------------------------

axes[0].plot(
    cnn_lstm_results["History"].history["accuracy"],
    label="CNN-LSTM Train"
)

axes[0].plot(
    cnn_lstm_results["History"].history["val_accuracy"],
    label="CNN-LSTM Validation"
)

axes[0].plot(
    cnn_gru_results["History"].history["accuracy"],
    label="CNN-GRU Train"
)

axes[0].plot(
    cnn_gru_results["History"].history["val_accuracy"],
    label="CNN-GRU Validation"
)

axes[0].set_title(
    "CNN-LSTM / CNN-GRU Accuracy"
)

axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")

axes[0].legend()
axes[0].grid(True)


# -----------------------------
# Loss
# -----------------------------

axes[1].plot(
    cnn_lstm_results["History"].history["loss"],
    label="CNN-LSTM Train"
)

axes[1].plot(
    cnn_lstm_results["History"].history["val_loss"],
    label="CNN-LSTM Validation"
)

axes[1].plot(
    cnn_gru_results["History"].history["loss"],
    label="CNN-GRU Train"
)

axes[1].plot(
    cnn_gru_results["History"].history["val_loss"],
    label="CNN-GRU Validation"
)

axes[1].set_title(
    "CNN-LSTM / CNN-GRU Loss"
)

axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")

axes[1].legend()
axes[1].grid(True)


plt.tight_layout()

plt.savefig(
    "/content/CNN_LSTM_GRU_Training_Curves.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 11. SECTION 25 CONSOLIDATED RESULTS
# ============================================================

final_results_table = pd.DataFrame([

    {
        "Model": "RNN",
        "Accuracy": rnn_results["Accuracy"],
        "Precision": rnn_results["Precision"],
        "Recall": rnn_results["Recall"],
        "F1": rnn_results["F1-Score"],
        "Parameters": rnn_results["Parameters"]
    },

    {
        "Model": "LSTM",
        "Accuracy": lstm_results["Accuracy"],
        "Precision": lstm_results["Precision"],
        "Recall": lstm_results["Recall"],
        "F1": lstm_results["F1-Score"],
        "Parameters": lstm_results["Parameters"]
    },

    {
        "Model": "GRU",
        "Accuracy": gru_results["Accuracy"],
        "Precision": gru_results["Precision"],
        "Recall": gru_results["Recall"],
        "F1": gru_results["F1-Score"],
        "Parameters": gru_results["Parameters"]
    },

    {
        "Model": "CNN-LSTM",
        "Accuracy": cnn_lstm_results["Accuracy"],
        "Precision": cnn_lstm_results["Precision"],
        "Recall": cnn_lstm_results["Recall"],
        "F1": cnn_lstm_results["F1-Score"],
        "Parameters": cnn_lstm_results["Parameters"]
    },

    {
        "Model": "CNN-GRU",
        "Accuracy": cnn_gru_results["Accuracy"],
        "Precision": cnn_gru_results["Precision"],
        "Recall": cnn_gru_results["Recall"],
        "F1": cnn_gru_results["F1-Score"],
        "Parameters": cnn_gru_results["Parameters"]
    }
])


# ============================================================
# 12. DISPLAY FINAL TABLE
# ============================================================

print("\n" + "=" * 70)
print("SECTION 25 — CONSOLIDATED RESULTS")
print("=" * 70)

display(
    final_results_table.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1": "{:.4f}",
        "Parameters": "{:,}"
    })
)


# ============================================================
# 13. SAVE FINAL TABLE
# ============================================================

final_results_table.to_csv(
    "/content/Section_25_Consolidated_Results.csv",
    index=False
)

print("\nFinal consolidated table saved to:")

print(
    "/content/Section_25_Consolidated_Results.csv"
)


print("\n" + "=" * 70)
print("PART 6 COMPLETED")
print("=" * 70)
```
