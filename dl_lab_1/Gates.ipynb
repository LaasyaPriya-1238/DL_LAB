```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class Perceptron:

    def __init__(self, lr=0.1, epochs=10):
        self.lr = lr
        self.epochs = epochs
        self.weights = None
        self.bias = 0

    def activation(self, x):
        if x >= 0:
            return 1
        return 0
```

```python
def plot_boundary(X, y, weights, bias, title, save=False):

    plt.figure(figsize=(5,5))

    for i in range(len(X)):
        if y[i] == 0:
            plt.scatter(X[i][0], X[i][1], color='red', s=100)
        else:
            plt.scatter(X[i][0], X[i][1], color='blue', s=100)

    x_values = np.linspace(-0.5, 1.5, 100)

    if weights[1] != 0:
        y_values = -(weights[0]*x_values + bias)/weights[1]
        plt.plot(x_values, y_values, color='black', linewidth=2)

    plt.xlim(-0.5, 1.5)
    plt.ylim(-0.5, 1.5)

    plt.grid(True)
    plt.title(title)

    if save:
        plt.savefig(title + ".png", dpi=300, bbox_inches="tight")

    plt.show()
    plt.close()
```

```python
def train(model, X, y, gate):

    model.weights = np.zeros(X.shape[1])
    model.bias = 0

    update = 1

    for epoch in range(model.epochs):

        print("\nEpoch", epoch + 1)

        error_found = False

        for i in range(len(X)):

            net = np.dot(X[i], model.weights) + model.bias

            prediction = model.activation(net)

            error = y[i] - prediction

            if error != 0:

                model.weights += model.lr * error * X[i]
                model.bias += model.lr * error

                print("---------------------------")
                print("Update :", update)
                print("Input :", X[i])
                print("Target :", y[i])
                print("Prediction :", prediction)
                print("Weights :", model.weights)
                print("Bias :", model.bias)

                plot_boundary(
                    X,
                    y,
                    model.weights,
                    model.bias,
                    gate + " Update " + str(update),
                    save=True
                )

                update += 1
                error_found = True

        if not error_found:
            print("\nTraining Converged")
            break
```

```python
def save_boundary(X, y, weights, bias, filename):

    plt.figure(figsize=(5,5))

    for i in range(len(X)):
        if y[i] == 0:
            plt.scatter(X[i][0], X[i][1], color='red', s=100)
        else:
            plt.scatter(X[i][0], X[i][1], color='blue', s=100)

    x_values = np.linspace(-0.5, 1.5, 100)

    if weights[1] != 0:
        y_values = -(weights[0]*x_values + bias)/weights[1]
        plt.plot(x_values, y_values, color='black', linewidth=2)

    plt.xlim(-0.5,1.5)
    plt.ylim(-0.5,1.5)

    plt.grid(True)
    plt.title(filename)

    plt.savefig(filename + ".png", dpi=300, bbox_inches="tight")

    plt.show()

    plt.close()
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0,0,0,1])

model = Perceptron()

train(model, X, y, "AND")

print("\nFinal Weights :", model.weights)
print("Final Bias :", model.bias)

save_boundary(
    X,
    y,
    model.weights,
    model.bias,
    "AND_DecisionBoundary"
)
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0,1,1,0])

model = Perceptron()

# Disable automatic image saving temporarily
original_plot = plot_boundary

def xor_plot_boundary(X, y, weights, bias, title, save=False):
    original_plot(X, y, weights, bias, title, save=False)

plot_boundary = xor_plot_boundary

# Train XOR
train(model, X, y, "XOR")

print("\nFinal Weights :", model.weights)
print("Final Bias :", model.bias)

# Save only the final decision boundary
save_boundary(
    X,
    y,
    model.weights,
    model.bias,
    "XOR_DecisionBoundary"
)

# Restore the original function
plot_boundary = original_plot
```
